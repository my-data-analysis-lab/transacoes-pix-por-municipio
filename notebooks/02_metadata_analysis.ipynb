# Analisando os metadados DataFrame.index

.index — índice/rótulos das linhas.

.columns — nomes/rótulos das colunas.

.dtypes — tipos de dados de cada coluna.

.info() — resumo geral do DataFrame.

.axes — retorna os eixos do DataFrame: linhas e colunas.

.ndim — quantidade de dimensões do DataFrame.

.size — quantidade total de elementos/células.

.shape — quantidade de linhas e colunas.

.memory_usage() — memória utilizada pelas colunas.

.empty — indica se o DataFrame está vazio.
