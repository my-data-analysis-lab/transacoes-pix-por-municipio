# Extração dos dados - request lib

```python
import pandas as pd
import requests
from requests.exceptions import RequestException
from requests.models import Response

try:
    response: Response = requests.get("https://olinda.bcb.gov.br/olinda/servico/Pix_DadosAbertos/versao/v1/odata/TransacoesPixPorMunicipio(DataBase=@DataBase)?@DataBase=%27202011%27&$top=100&$format=json")
except RequestException as e:
    raise RequestException(f"Erro: {e}")
else:
    data = response.json()['value']

    if data:
        csv = pd.DataFrame(data)
        csv.to_csv('../data/raw/pix.csv')
```
